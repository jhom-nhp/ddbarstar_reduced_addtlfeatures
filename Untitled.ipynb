```python
import qrcode

# The text or URL you want the QR code to point to
data = "https://www.facebook.com/profile.php?id=100086340947226"

# Generate the QR code image
img = qrcode.make(data)

# Save the image file to your current directory
img.save("basic_qrcode.png")

print("QR code successfully saved as 'basic_qrcode.png'")
```
